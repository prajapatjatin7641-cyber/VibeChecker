```python
#3.
for i in range(1, 6):
    for j in range(1, i+1):
        if i%2==0:
            if j%2==0:
                print("1", end=" ")
            else:
                print("0", end=" ")
        else:
            if j%2==0:
                print("0", end=" ")
            else:
                print("1", end=" ")
    print()
```

```python
#4.
for i in range(0, 5):
    for j in range(0, i):
        print(" ", end="")
    for j in range(0, 9-i*2):
        print("*", end="")
    print()
```
